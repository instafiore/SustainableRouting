```python
import os
from typing import Dict
import pandas as pd
import re
import ast
import matplotlib.pyplot as plt

def plot_stats(file, metrics = ["Occupancy", "OccupancyWatched"], old = False, controlledRangesInput = []):
    r = r"stats: (\{.*\})"

    # df = pd.DataFrame(columns=["TimeLoss", "DepartDelay", "AccumulatedWaitingTime", "Occupancy"])
    df = pd.DataFrame(columns = metrics)
    # df = pd.DataFrame(columns = ["OccupancyWatched"])

    regex_map = r"mn.(\w+)"

    map_capacity = {
        "acosta_buslanes": CAPACITY_BOLOGNA,
        "net": CAPACITY_MK,
        "subnetLust": CAPACITY_LUX,
    }

    _map = re.search(regex_map, file).group(1)
    capacity = float(map_capacity[_map])

    controlledRanges = []

    with open(file=file, mode="r") as f:
        lines = f.readlines()
        controlling = True
        for l in lines:
            m = re.search(r, l)
            if not m: continue
            s = m.group(1)
            stats: Dict = ast.literal_eval(s)
            if old:
                numVehicles = int(stats["#Vehicles"])
                stats["capacity"] = capacity
                stats["Occupancy"] = numVehicles/capacity
            i = len(df)
            df.loc[i] = stats
            if re.search(r"Controlled: False", l):
                if controlling:
                    controlling = False
                    controlledRanges.append(i)
            else:
                if not controlling:
                    controlling = True
                    controlledRanges.append(i)
        controlledRanges.append(len(df))
    columns = df.columns

    # fig, axes = plt.subplots(len(columns), ncols=1, figsize=(14, 8), sharex=True)
    plt.tight_layout(pad=5.0)
    fig, ax = plt.subplots(1,figsize=(7, 3), sharex=True)
    plt.minorticks_on()
    # axes = [axes] if len(columns) == 1 else axes

    ax.set_title(f"Map {_map}")
    ax.grid(True)
    ax.set_ylabel("Value")
    ax.set_xlabel("#Vehicles")
    ax.set_ylim(0,1)
    print(controlledRanges)
    for vertical_line in controlledRanges:
        plt.axvline(x=vertical_line, color='red', linestyle='--', linewidth=2)

    for vertical_line in controlledRangesInput:
        plt.axvline(x=vertical_line, color='green', linestyle='--', linewidth=2)

    for i in range(len(columns)):
        # ax = axes[i]
        column = columns[i]
        ax.plot(df.index, df[column])
        
    
    print(df.head())
```
